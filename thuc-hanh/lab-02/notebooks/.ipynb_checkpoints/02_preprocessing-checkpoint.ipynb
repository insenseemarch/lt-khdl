# 02_Preprocessing.ipynb

## **0. Giới thiệu**

Notebook này thực hiện tiền xử lý dữ liệu theo đúng yêu cầu bài lab:

- Làm sạch và chuẩn hoá dữ liệu.

- Xử lý giá trị không hợp lệ.

- Xử lý missing values.

- Phát hiện và loại bỏ/điều chỉnh outlier.

- Chuẩn hoá dữ liệu.

- Feature engineering cơ bản.

## **1. Import thư viện và load dữ liệu từ file `01_data_exploration.ipynb`**

```python
import numpy as np
import sys, os

project_root = os.path.abspath("..")
sys.path.append(project_root)


from src.data_processing import *


hdr, data = load_csv_numpy("../data/raw/aug_test.csv")
y = np.load("../data/raw/jobchange_test_target_values.npy")
if data.shape[0] != len(y):
    raise RuntimeError("Số dòng csv và target npy không khớp")
data = np.hstack([data, y.reshape(-1,1)])
hdr = np.append(hdr, "target")
print("Loaded:", data.shape)
```

## **2. Xử lý giá trị không hợp lệ**

### Xử lý experience (<1, >20)

```python
def clean_special_values(data, headers):
    h = list(headers)

    # Experience
    if "experience" in h:
        i = h.index("experience")
        col = np.char.strip(data[:, i].astype(str))
        col = np.where(col == "<1", "0", col)
        col = np.where(col == ">20", "21", col)
        col = np.where(col == "", "nan", col)
        data[:, i] = col

    # Last_new_job
    if "last_new_job" in h:
        i = h.index("last_new_job")
        col = np.char.strip(data[:, i].astype(str))
        col = np.where(col == "never", "0", col)
        col = np.where(col == ">4", "5", col)
        col = np.where(col == "", "nan", col)
        data[:, i] = col

clean_special_values(data, hdr)
```

## **3. Xử lý Missing Values**

### 3.1 Numeric: fill bằng median

```python
numeric_cols = ["city_development_index", "experience", "training_hours"]
hlist = list(hdr)

for name in numeric_cols:
    idx = hlist.index(name)
    arr = to_numeric_col(data[:, idx], fill_nan_with=np.nan)
    med = np.nanmedian(arr)
    data[:, idx] = np.where(np.isnan(arr), med, arr).astype(object)
```

### 3.2 Categorical: fill bằng mode

```python
for j, name in enumerate(hdr):
    if name in numeric_cols or name in ("target", "enrollee_id"):
        continue

    col = data[:, j].astype(str)
    missing = missing_mask(col)
    if (~missing).sum() == 0:
        fillv = ""
    else:
        fillv = mode_np(col[~missing])

    data[:, j] = np.where(missing, fillv, col)
```

## **4. Phát hiện & xử lý Outlier**

```python
cat_cols = [h for h in hdr if h not in numeric_cols + ["target", "enrollee_id"]]
enc_mapping = {}

for name in cat_cols:
    idx = hdr.tolist().index(name)
    encoded, uniques = label_encode_np(data[:, idx].astype(str))
    data[:, idx] = encoded.astype(object)
    enc_mapping[name] = uniques.tolist()
```

```python
for name in numeric_cols:
    idx = hdr.tolist().index(name)
    num = to_numeric_col(data[:, idx])
    clipped = clip_iqr_np(num)
    data[:, idx] = clipped.astype(object)
```

## **5. Chuẩn hoá dữ liệu**

```python
for name in numeric_cols:
    idx = hdr.tolist().index(name)
    arr = to_numeric_col(data[:, idx])
    
    if name == "experience":
        data[:, idx] = arr.astype(object)
    else:
        data[:, idx] = minmax_np(arr).astype(object)
```

## **6. Feature Engineering**

Tương tác CDI × training_hours tạo thêm thông tin mới:
ứng viên ở thành phố phát triển + được training nhiều → khả năng đổi việc khác nhóm còn lại.

```python
i_cdi = hdr.tolist().index("city_development_index")
i_tr  = hdr.tolist().index("training_hours")

cdi = to_numeric_col(data[:, i_cdi])
th  = to_numeric_col(data[:, i_tr])

new_feature = (cdi * th).reshape(-1, 1)

data = np.hstack([data, new_feature.astype(object)])
headers = np.append(hdr, "cdi_x_training")
```

Dữ liệu thực tế cho thấy người 1–4 năm là nhóm nhảy việc nhiều nhất (theo phân tích ở file `01_data_exploration.ipynb`). Binning làm mô hình học “nhóm hành vi” tốt hơn so với số năm thô.

```python
idx_exp = headers.tolist().index("experience")

exp = to_numeric_col(data[:, idx_exp])
bins = [0, 1, 4, 9, 20, 100]
exp_bin = np.digitize(exp, bins)
data = np.hstack([data, exp_bin.reshape(-1,1)])
headers = np.append(hdr, "experience_level")
```

Trong phân tích, người chưa bao giờ đổi job trước đây (`never`) lại là nhóm muốn đổi nhiều nhất hiện tại. Đây là insight rất quan trọng, cần feature engineering để mô hình học được tương quan này.

```python
idx_last = headers.tolist().index("last_new_job")

lj = np.char.strip(data[:, idx_last].astype(str))

lj = np.where(lj == "0", "1", lj)
lj = np.where(lj == "1", "2", lj)
lj = np.where(lj == "2", "3", lj)
lj = np.where(lj == "3", "4", lj)
lj = np.where(lj == "4", "5", lj)
lj = np.where(lj == "5", "0", lj)   # nhóm >4 xem như mới đổi

stability = lj.astype(float).astype(int)

data = np.hstack([data, stability.reshape(-1,1)])
headers = np.append(headers, "job_stability_score")
```

## **7. Lưu dữ liệu đã tiền xử lý**

```python
ndata = np.empty(data.shape, dtype=float)
for j in range(data.shape[1]):
    ndata[:, j] = to_numeric_col(data[:, j], fill_nan_with=np.nan)

ndata = np.nan_to_num(ndata, nan=0)

final_headers = [h for h in headers if h not in ["target", "enrollee_id"]]

np.save("../data/processed/data_test_clean.npy", ndata)
np.save("../data/processed/headers_test_clean.npy", np.array(final_headers))
```
