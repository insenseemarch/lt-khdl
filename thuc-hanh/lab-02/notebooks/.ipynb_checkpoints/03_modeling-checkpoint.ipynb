## **1. Import & Load dữ liệu đã tiền xử lý**

```python
import numpy as np, os, sys

project_root = os.path.abspath("..")
sys.path.append(project_root)

from src.models import *
from src.data_processing import *
from src.visualization import *

data = np.load("../data/processed/data_test_clean.npy", allow_pickle=True)
headers = np.load("../data/processed/headers_test_clean.npy", allow_pickle=True).astype(str)
y = np.load("../data/raw/jobchange_test_target_values.npy")

print("data shape:", data.shape)
print()

print("headers:", headers.tolist())
print()

print("y shape:", y.shape)
print()

print(data.shape, headers)
```

## **2. Xác định biến đầu vào & target**

```python
hdr = headers.tolist()

drop_cols = ["enrollee_id", "target"]
feat_headers = [h for h in hdr if h not in drop_cols]

keep_idx = [int(np.where(headers == h)[0][0]) for h in feat_headers]

X = data[:, keep_idx].astype(float)   # shape (n, d)
if X.shape[0] != y.shape[0]:
    raise RuntimeError(f"Row mismatch: X has {X.shape[0]} rows but y has {y.shape[0]} rows")

print("X shape:", X.shape, "y shape:", y.shape)
print()
print("feature headers:", feat_headers)
```

## **3. Train-test split**

```python
n = X.shape[0]
rng = np.random.RandomState(0)
idx = rng.permutation(n)
cut1 = int(0.8 * n)
cut2 = int(0.9 * n)
train_idx = idx[:cut1]; val_idx = idx[cut1:cut2]; test_idx = idx[cut2:]

X_train, y_train = X[train_idx], y[train_idx].astype(int)
X_val,   y_val   = X[val_idx],   y[val_idx].astype(int)
X_test,  y_test  = X[test_idx],  y[test_idx].astype(int)

train_mean = X_train.mean(axis=0)
train_std  = X_train.std(axis=0)
train_std[train_std == 0] = 1.0
X_train = (X_train - train_mean) / train_std
X_val   = (X_val   - train_mean) / train_std
X_test  = (X_test  - train_mean) / train_std

print("shapes (train/val/test):", X_train.shape, X_val.shape, X_test.shape)
```

## **4. Mô hình Logistic Regression**

```python
w, b, loss_hist = logistic_regression_sgd(
    X_train, y_train,
    lr=0.1, epochs=200,
    batch_size=128, l2=1e-3,
    verbose=False
)

yhat_test = predict_label(X_test, w, b)
print("NumPy Logistic - Test accuracy:", accuracy(y_test, yhat_test))

metrics = precision_recall_f1(y_test, yhat_test)

print("Precision:", metrics["precision"])
print()

print("Recall:",    metrics["recall"])
print()

print("F1-score:",  metrics["f1"])
print()

print("Confusion matrix:\n", metrics["confusion"])
```

## **5. Mô hình KNN**

```python
def knn_predict(X_train, y_train, X_test, k=5):
    preds = []
    for x in X_test:
        # khoảng cách Euclidean
        dists = np.sqrt(((X_train - x) ** 2).sum(axis=1))
        idx = np.argsort(dists)[:k]
        votes = y_train[idx]
        pred = 1 if votes.mean() >= 0.5 else 0
        preds.append(pred)
    return np.array(preds)

yhat_knn = knn_predict(X_train, y_train, X_test, k=5)

print("KNN Accuracy:", accuracy(y_test, yhat_knn))
print()

m_knn = precision_recall_f1(y_test, yhat_knn)
print("KNN Precision:", m_knn["precision"])
print()

print("KNN Recall:", m_knn["recall"])
print()

print("KNN F1:", m_knn["f1"])
print()

print("KNN Confusion:\n", m_knn["confusion"])
```

## **6. Visualization và nhận xét mô hình**

### 6.1. Loss Curve

```python
plt.figure(figsize=(8,4))
plt.plot(loss_hist)
plt.title("Hàm mất mát (Loss) qua các epoch", fontsize=14)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True)
plt.show()
```

Nhận xét:

- Đường loss giảm rất nhanh trong khoảng **0–25 epoch**, nghĩa là mô hình học được pattern chính ngay từ đầu.

- Sau ~50 epoch, loss gần như đã **hội tụ**, dao động nhẹ quanh một giá trị ổn định.

Điều này cho thấy tốc độ học của Logistic Regression ổn, learning rate 0.1 dùng trong mô hình **không gây overshoot**, và mô hình không bị diverge.

**Kết luận**: Quá trình tối ưu hội tụ tốt, mô hình học ổn định, không bị overfitting hay underfitting ở mức loss.

### 6.2. Precision–Recall curve

```python
proba = predict_proba(X_test, w, b)

precisions = []
recalls = []
thresholds = np.linspace(0, 1, 200)

for th in thresholds:
    pred = (proba >= th).astype(int)
    m = precision_recall_f1(y_test, pred)
    precisions.append(m["precision"])
    recalls.append(m["recall"])

plt.figure(figsize=(5,5))
plt.plot(recalls, precisions)
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve (NumPy)")
plt.grid(True)
plt.show()
```

Nhận xét:

- Precision duy trì ở mức **gần 1.0** trên hầu hết các ngưỡng threshold => mô hình **dự đoán nhãn 1 rất chính xác** (ít false positive).

- Recall tăng đều khi giảm threshold, đạt mức rất cao ở vùng 0.8–1.0 => mô hình bắt được đa phần các trường hợp thuộc lớp 1.

- Đồ thị Precision–Recall tương đối “vuông”, chứng tỏ mô hình cân bằng tốt giữa precision và recall.

**Kết luận**: Logistic Regression tự xây dựng đạt hiệu năng cao ở cả hai phía precision và recall nên rất tốt cho dữ liệu có tính mất cân bằng nhẹ giữa hai lớp.

### 6.3. Confusion Matrix Heatmap

```python
cm = confusion_matrix_np(y_test, yhat_test)

plt.figure(figsize=(4,3))
plt.imshow(cm, cmap="Blues")
plt.colorbar()

for i in range(2):
    for j in range(2):
        plt.text(j, i, cm[i,j], ha="center", va="center", color="black", fontsize=14)

plt.xticks([0,1], ["Pred 0", "Pred 1"])
plt.yticks([0,1], ["True 0", "True 1"])
plt.title("Confusion Matrix - NumPy Logistic")
plt.show()
```

Nhận xét:

- TN = 154, TP = 59, FP = 0, FN = 0

- Không có **false positive và false negative** nào.

- Mô hình phân loại **chính xác tuyệt đối** trên tập test (accuracy = 1.0).

Điều này cho thấy:

- Logistic Regression hoạt động cực tốt trên bộ dữ liệu đã chuẩn hoá và tiền xử lý.

- Các đặc trưng sau preprocessing (đặc biệt feature engineering như `cdi × training` và binning experience) giúp mô hình tách biệt 2 lớp rất rõ ràng.

**Kết luận**: Mô hình Logistic đạt hiệu năng hoàn hảo trên tập test.

### 6.4. So sánh Model (LR vs KNN)

```python
acc_lr  = accuracy(y_test, yhat_test)
acc_knn = accuracy(y_test, yhat_knn)

plt.figure(figsize=(5,4))
plt.bar(["LR (NumPy)", "KNN (NumPy)"], [acc_lr, acc_knn])
plt.ylabel("Accuracy")
plt.title("So sánh độ chính xác giữa các mô hình")
plt.ylim(0,1)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```

Nhận xét:

- Logistic Regression (NumPy) và KNN (NumPy) đều đạt **accuracy ≈ 1.0**.

- Tuy nhiên, Logistic Regression ổn định hơn vì:

    - KNN không học được trọng số => phụ thuộc vào khoảng cách.

    - Logistic có hàm tối ưu rõ ràng => ít bị nhiễu.

- Với dữ liệu numeric đã chuẩn hoá, Logistic Regression luôn hoạt động tốt hơn hoặc ổn định hơn KNN.

**Kết luận**: Hai mô hình đều mạnh, nhưng **Logistic Regression** có độ tin cậy và khả năng tổng quát hoá **tốt hơn**.
