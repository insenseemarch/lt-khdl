Notebook này thực hiện khám phá dữ liệu thô trên bộ dữ liệu [HR Analytics: Job Change of Data Scientists](https://www.kaggle.com/datasets/arashnic/hr-analytics-job-change-of-data-scientists?select=aug_train.csv). Mục tiêu:

## **1. Import thư viện & Load dữ liệu thô**

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys, os

project_root = os.path.abspath("..")
sys.path.append(project_root)

from src.data_processing import *
from src.visualization import *

sns.set(style='whitegrid', rc={'figure.figsize': (8,5)})
```

```python
headers, data = load_csv_numpy('../data/raw/aug_test.csv')
targets = np.load("../data/raw/jobchange_test_target_values.npy")

data = np.hstack([data, targets.reshape(-1,1)])
headers = np.append(headers, "target")

print("headers:")
for h in headers:
    print(f'- {h}')
    
print("\nshape:", data.shape)
```

```python
# in 5 dòng đầu

for r in data[:5]:
    print(", ".join(map(str, r)))
```

## **2. Kiểm tra missing values theo từng cột**

```python
total = data.shape[0]

miss_counts = np.array([ missing_mask(data[:,i]).sum() for i in range(data.shape[1]) ])
miss_pct = miss_counts / total * 100

for h, cnt, pct in zip(headers, miss_counts, miss_pct):
    print(f"{h:25s}: {cnt:5d} ({pct:5.2f}%)")
```

## **3. Phân loại cột theo kiểu dữ liệu (numeric / categorical)**

```python
explicit_numeric = ["city_development_index", "experience", "training_hours"]
numeric_cols = []
categorical_cols = []

for i, h in enumerate(headers):
    if h in explicit_numeric:
        numeric_cols.append(h)
    elif h in ("enrollee_id","target"):
        # 2 cột này không có ý nghĩa trong các xử lý sau
        continue
    else:
        col = data[:,i].astype(str)
        mask = ~missing_mask(col)
        vals = col[mask]
        if vals.size==0:
            categorical_cols.append(h)
        else:
            num_ok = 0
            for v in vals:
                try:
                    float(v)
                    num_ok += 1
                except:
                    pass
            if num_ok / vals.size >= 0.9:
                numeric_cols.append(h)
            else:
                categorical_cols.append(h)

print("numeric:")
for i in numeric_cols:
    print(f'- {i}')
print()
print("categorical:")
for i in categorical_cols:
    print(f'- {i}')
```

## **4. Thống kê mô tả cho biến numeric**

```python
# xử lý tạm thời các dấu '<', '>'

for h in numeric_cols:
    idx = list(headers).index(h)

    if h == "experience":
        # dùng hàm mới
        col = to_numeric_col(data[:, idx])
        arr = col[~np.isnan(col)]
    else:
        col = to_numeric_col(data[:, idx])
        arr = col[~np.isnan(col)]

    if arr.size==0:
        print(h, "no data")
    else:
        print(f"--- {h} ---")
        print(f'- count: {arr.size}')
        print(f'- min: {arr.min()}')
        print(f'- max: {arr.max()}')
        print(f'- mean: {arr.mean()}')
        print(f'- median: {np.median(arr)}')
        print(f'- std: {arr.std()}')
        print()
```

```python
idx = headers.tolist().index("training_hours")
col_train = data[:, idx].astype(str)
col_train = col_train[~missing_mask(col_train)].astype(float)

plot_hist_numeric(col_train, "Training Hours (Histogram - 5 bins)")
```

## **5. Thống kê cho biến categorical (Top 5 giá trị phổ biến)**

```python
TOP_K = 5

for h in categorical_cols:
    idx = list(headers).index(h)
    col = data[:,idx].astype(str)
    mask = ~missing_mask(col)
    vals = col[mask]
    if vals.size==0:
        continue
    uniq, inv = np.unique(vals, return_inverse=True)
    counts = np.bincount(inv)
    order = np.argsort(-counts)
    top = order[:min(TOP_K, len(order))]
    total = vals.size
    print("---", h, "---")
    for rank, j in enumerate(top, start=1):
        print(f"{rank}. {uniq[j]} — {counts[j]} ({counts[j]/total*100:.2f}%)")
    print()
```

**Tiếp tục mô hình hóa dữ liệu để tìm hiểu thêm một số thông tin của các ứng viên trong ngành Khoa học dữ liệu.**

```python
col = data[:, headers.tolist().index("city")]
plot_top_categories(col, N=5, title="5 thành phố có lượng nhân viên Khoa Học Dữ Liệu nhiều nhất")
```

```python
idx = headers.tolist().index("education_level")
col_edu = data[:, idx].astype(str)
col_edu = col_edu[~missing_mask(col_edu)]
uniq, inv = np.unique(col_edu, return_inverse=True)
counts = np.bincount(inv)

order = np.argsort(-counts)
uniq_sorted = uniq[order]
counts_sorted = counts[order]

plt.figure(figsize=(7,4))
sns.barplot(x=counts_sorted, y=uniq_sorted)
plt.title("Phổ biến các mức Trình độ học vấn")
plt.xlabel("Số lượng")
plt.ylabel("Trình độ học vấn")
plt.show()
```

## **6. Phát hiện Outlier (chỉ nhận diện)**

```python
def iqr_stats(arr):
    # arr: 1D numeric (no NaN)
    if arr.size == 0:
        return 0, np.nan, np.nan
    q1 = np.percentile(arr, 25)
    q3 = np.percentile(arr, 75)
    iqr = q3 - q1
    low = q1 - 1.5 * iqr
    high = q3 + 1.5 * iqr
    out_cnt = int(((arr < low) | (arr > high)).sum())
    return out_cnt, low, high

for h in numeric_cols:
    idx = list(headers).index(h)

    # experience đã được clean -> chỉ convert sang numeric
    colnum = to_numeric_col(data[:, idx], fill_nan_with=np.nan)
    arr = colnum[~np.isnan(colnum)]

    out_cnt, lb, ub = iqr_stats(arr)
    print("---", h, "---")
    print(f'- outliers: {out_cnt}')
    print(f'- rate: {out_cnt/arr.size*100:.2f}%')
    print(f'- bounds: {lb}, {ub}')
    print()
```

### 7.2. Nhóm nào thường có xu hướng muốn đổi việc? (giới tính, kinh nghiệm, học vấn)

```python
gender = data[:, headers.tolist().index("gender")]
target = data[:, headers.tolist().index("target")].astype(int)

exp = to_numeric_col(data[:, headers.tolist().index("experience")])
edu = data[:, headers.tolist().index("education_level")]

mask_change = target == 1

gender_c = gender[mask_change]
exp_c     = exp[mask_change]
edu_c     = edu[mask_change]

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

colors = ["#5DADE2", "#48C9B0", "#F5B041", "#AF7AC5", "#EC7063"]

# =========================
# PLOT 1: Gender
# =========================
uniq_g, counts_g = np.unique(gender_c, return_counts=True)
perc_g = counts_g / counts_g.sum() * 100

ax = axes[0]
bars = ax.bar(uniq_g, perc_g, color=colors[:len(uniq_g)])
ax.set_title("Tỉ lệ đổi việc theo giới tính")
ax.set_ylabel("Tỉ lệ (%)")

for bar in bars:
    h = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2, h + 1, f"{h:.1f}%", ha="center")

# =========================
# PLOT 2: Experience
# =========================
exp_clean = exp_c[~np.isnan(exp_c)]

bins = [0, 1, 5, 10, 20, 50]
labels = ["<1", "1-4", "5-9", "10-19", ">20"]

exp_binned = np.digitize(exp_clean, bins)

vals2 = []
rates2 = []

for i, lb in enumerate(labels, start=1):
    mask = exp_binned == i
    if mask.sum() > 0:
        vals2.append(lb)
        rates2.append(mask.sum())

rates2 = np.array(rates2)
rates2 = rates2 / rates2.sum() * 100

ax = axes[1]
bars = ax.bar(vals2, rates2, color=colors[:len(vals2)])
ax.set_title("Tỉ lệ đổi việc theo kinh nghiệm")
ax.set_ylabel("Tỉ lệ (%)")

for bar in bars:
    h = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2, h + 1, f"{h:.1f}%", ha="center")

# =========================
# PLOT 3: Education level
# =========================
uniq_e, counts_e = np.unique(edu_c, return_counts=True)
perc_e = counts_e / counts_e.sum() * 100

ax = axes[2]
bars = ax.bar(uniq_e, perc_e, color=colors[:len(uniq_e)])
ax.set_title("Tỉ lệ đổi việc theo trình độ học vấn")
ax.set_ylabel("Tỉ lệ (%)")
ax.set_xticks(np.arange(len(uniq_e)))
ax.set_xticklabels(uniq_e, rotation=45, ha="right")

for bar in bars:
    h = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2, h + 1, f"{h:.1f}%", ha="center")

plt.tight_layout()
plt.show()
```

**Ghi chú**: Những dòng bị missing value sẽ được gom về một cột không có tên.

Như vậy, ta xét từng biểu đồ một:
- Tỉ lệ đổi việc theo giới tính: Nhóm Male chiếm phần lớn trong số những người muốn đổi việc (~63%). Female và Other chiếm tỉ lệ khá thấp trong nhóm này. Điều này có thể dễ dàng được lý giải bởi tỉ lệ nam chiếm 90.1% trong công việc này.
- Tỉ lệ đổi việc theo kinh nghiệm:
    - Nhóm 1–4 năm kinh nghiệm dẫn đầu về tỉ lệ đổi việc (~35%).
    - Nhóm 5–9 năm cũng có xu hướng cao (~27%).
    - Người mới vào nghề (<1 năm) ít đổi hơn - có thể do họ vừa ổn định công việc hoặc chưa đủ động lực để nhảy việc.
    - Nhóm 10+ năm giảm dần - thường ổn định hoặc chuyển sang vai trò cấp cao nên tần suất nhảy việc thấp hơn.
  
  => Nhìn chung, người ở khoảng middle chính là nhóm muốn đổi việc nhất trong ngành.

- Tỉ lệ đổi việc theo trình độ học vấn:
    - Graduate (Đại học) chiếm tỉ lệ cao nhất trong nhóm đổi việc (~66%).
    - Nhóm Master đứng thứ hai (~20%).
    - High School và PhD cực thấp.
    
    => Điều này khá hợp lý: lực lượng lao động chính của ngành Khoa học Dữ liệu là cử nhân, và chính họ là nhóm am hiểu nhất khi cơ hội thị trường thay đổi.

Vậy ta tiếp tục xem xét trong nhóm Cử nhân, việc ứng viên học chuyên ngành nào sẽ có tỉ lệ muốn đổi việc Khoa học dữ liệu cao nhất, để xem liệu những người học trái ngành có cảm thấy 'hối hận' khi chọn công việc này hay không.

### 7.3. Với trình độ học vấn Cử nhân, chuyên ngành nào dễ đổi việc nhất?

```python
edu = data[:, headers.tolist().index("education_level")]
major = data[:, headers.tolist().index("major_discipline")]
target = data[:, headers.tolist().index("target")].astype(int)

# Graduate's mask
mask_grad = edu == "Graduate"
major_grad = major[mask_grad]
target_grad = target[mask_grad]

uniq, counts = np.unique(major_grad, return_counts=True)

vals = []
rates = []

MIN_COUNT = 10   # eliminate under-10 groups

for val, cnt in zip(uniq, counts):
    if cnt < MIN_COUNT:
        continue
    mask = major_grad == val
    vals.append(val)
    rates.append(target_grad[mask].mean() * 100)

colors = ["#5DADE2", "#48C9B0", "#F5B041", "#AF7AC5", "#EC7063", "#5499C7"]

plt.figure(figsize=(10,5))
bars = plt.bar(vals, rates, color=colors[:len(vals)])

plt.title("Tỉ lệ đổi việc theo chuyên ngành – Nhóm Graduate")
plt.ylabel("Tỉ lệ đổi việc (%)")
plt.xticks(rotation=45, ha="right")

for bar in bars:
    h = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, h + 1, f"{h:.1f}%", ha="center")

plt.tight_layout()
plt.show()
```

Dễ dàng nhận thấy:

- Business Degree nhảy việc mạnh nhất (~34.8%). Lý do hợp lý:
    - Nhiều bạn từ ngành Business chuyển sang Data/Analytics khá phổ biến.
    - Khi vào làm Data Science, họ dễ có cảm giác “skill gap” do không được đào tạo thuần về dữ liệu, muốn tìm nơi khác training tốt hơn hoặc phù hợp hơn.
      
- Ngành STEM cũng có tỉ lệ đổi việc cao (~31%). Lý do:
    - Thị trường DS/ML cực năng động.
    - Người có nền tảng STEM rất dễ nhảy việc vì có nhiều lựa chọn hấp dẫn hơn (ML Engineer, Data Engineer, Research…)
    - Họ có kỹ năng tốt, tự tin hơn trong việc kiếm việc mới.


- Nhóm Humanities và Arts có tỉ lệ đổi việc thấp: Các nhóm này ít làm trái ngành vào Data Science nên những ai đã cố gắng vào được ngành thì yếu tố ổn định công việc và động lực giữ việc sẽ cao hơn.

- Nhóm “No Major” có tỉ lệ thấp nhất (~15%). Nhóm này thường đến từ self-learning. Họ ít đổi việc vì:
    - Khó tìm job mới hơn.
    - Họ cảm thấy công việc hiện tại đã phù hợp hoặc đủ tốt.

### 7.4. Người từng đổi job gần đây (last_new_job) có xu hướng muốn đổi nữa không?

```python
col_last = data[:, headers.tolist().index("last_new_job")].astype(str)
col_last = np.char.strip(col_last)
col_last = np.where(col_last == "", "Unknown", col_last)

col_target = data[:, headers.tolist().index("target")].astype(int)

uniq = np.unique(col_last)

ratios = []
counts = []

for u in uniq:
    mask = (col_last == u)
    total = np.sum(mask)
    pos = np.sum(col_target[mask] == 1)
    ratio = pos / total if total else 0
    ratios.append(ratio)
    counts.append(total)

# sort in descending order
order = np.argsort(-np.array(ratios))
uniq = uniq[order]
ratios = np.array(ratios)[order]

plt.figure(figsize=(8,4))
plt.bar(uniq, ratios)
plt.xticks(rotation=45, ha="right")
plt.ylabel("Tỷ lệ muốn đổi job (%)")
plt.title("Tỉ lệ muốn đổi việc của những ứng viên từng đổi việc trong quá khứ")
plt.tight_layout()
plt.show()
```

- Người chưa từng đổi job trước đây (never) lại có tỷ lệ muốn đổi job hiện tại cao nhất (~35%). Có thể vì họ đã gắn bó quá lâu với một nơi và bắt đầu muốn thay đổi môi trường. Hoặc họ đang cảm thấy công việc hiện tại không còn phù hợp với bản thân nữa.
- Người mới đổi việc 1–2 năm trước cũng có xu hướng muốn đổi nhưng thấp hơn nhóm `never`.
- Khi số năm kể từ lần đổi job gần nhất tăng lên 3–4 năm, tỷ lệ đổi job giảm dần đều.

Như vậy:
- Ứng viên càng ít thay đổi công việc trong quá khứ thì càng có xu hướng muốn đổi trong hiện tại.
- Người mới đổi việc gần đây lại không muốn đổi tiếp, có thể vì họ đang trong giai đoạn “ổn định lại”.

### 7.5. Những yếu tố nào là rủi ro cao dẫn đến việc ứng viên có thể đổi việc?

```python
rel_exp = data[:, headers.tolist().index("relevent_experience")]
train_hours = data[:, headers.tolist().index("training_hours")].astype(float)
cdi = data[:, headers.tolist().index("city_development_index")].astype(float)
target = data[:, headers.tolist().index("target")].astype(int)

fig, axes = plt.subplots(1, 3, figsize=(20, 5))

colors = ["#5DADE2", "#48C9B0", "#F5B041", "#AF7AC5", "#EC7063"]

# ======================================================
# Helper — plot the job-change rate for each group
# ======================================================
def plot_rate_group(ax, groups, target, title, xlabel, colors):
    uniq, counts = np.unique(groups, return_counts=True)
    
    rates = []
    for u in uniq:
        mask = (groups == u)
        change_rate = target[mask].mean() * 100
        rates.append(change_rate)

    ax.bar(uniq, rates, color=colors[:len(uniq)])
    ax.set_title(title)
    ax.set_ylabel("Tỉ lệ đổi việc (%)")
    ax.set_xlabel(xlabel)

    ax.set_xticks(np.arange(len(uniq)))
    ax.set_xticklabels(uniq, rotation=30, ha="right")

    # annotate
    for i, v in enumerate(rates):
        ax.text(i, v + 1, f"{v:.1f}%", ha="center")

# ======================================================
# 1) CDI — divided into Low / Medium / High
# ======================================================
cdi_bins = np.digitize(cdi, [0.6, 0.8])
cdi_labels = np.array(["Low (<0.6)", "Medium (0.6–0.8)", "High (>0.8)"])
cdi_groups = cdi_labels[cdi_bins]

plot_rate_group(
    axes[0], cdi_groups, target,
    "Tỉ lệ đổi việc theo City Development Index",
    "CDI Level", colors
)

# ======================================================
# 2) Training Hours
# ======================================================
tr_bins = np.digitize(train_hours, [20, 50, 100])
tr_labels = np.array(["0–20h", "21–50h", "51–100h", ">100h"])
tr_groups = tr_labels[tr_bins]

plot_rate_group(
    axes[1], tr_groups, target,
    "Tỉ lệ đổi việc theo số giờ Training",
    "Training Hours", colors
)

# ======================================================
# 3) Relevant Experience — Yes / No / Unknown
# ======================================================
rel_exp_clean = np.where(rel_exp == "", "Unknown", rel_exp)

plot_rate_group(
    axes[2], rel_exp_clean, target,
    "Tỉ lệ đổi việc theo Relevant Experience",
    "Relevant Experience", colors
)

plt.tight_layout()
plt.show()
```

1) City Development Index (CDI)
   
- Low CDI (<0.6): Tỉ lệ đổi việc **cao nhất ~60%**. Khu vực kém phát triển dễ khiến ứng viên cảm thấy thiếu cơ hội, thu nhập thấp, môi trường công việc hạn chế => **động lực đổi việc mạnh**.
  
- Medium CDI (0.6–0.8): Tỉ lệ **trung bình ~44%**. Cơ hội vừa phải => ứng viên **phân vân** giữa ở lại và rời đi.

- High CDI (>0.8): Tỉ lệ **thấp nhất ~18%**. Ứng viên ở khu vực phát triển thường **thỏa mãn hơn** với môi trường hiện tại.

**Kết luận**. CDI là yếu tố rủi ro số 1: sống ở khu vực kém phát triển làm tăng khả năng đổi việc.

2) Relevant Experience (Có kinh nghiệm làm việc trong ngành hay chưa?)
   
- No relevant experience: Tỉ lệ đổi việc **cao nhất ~37%**. Những ứng viên chưa từng làm việc trong ngành Khoa học dữ liệu thường **thiếu định hướng**, hoặc **chưa phù hợp** với job đầu tiên => dễ nhảy việc.
- 
- Has relevant experience: Tỉ lệ đổi việc **thấp hơn ~23%**. Ứng viên đã có kinh nghiệm ngành ổn định hơn, biết rõ kỳ vọng công việc.

**Kết luận**: Người thiếu kinh nghiệm ngành là nhóm dễ rời bỏ công việc nhất.

3) Training Hours – Số giờ training hiện tại
Các nhóm training đều có tỉ lệ đổi việc dao động rất gần nhau (~26–29%), nghĩa là:
- Training hiện tại không tạo khác biệt rõ rệt về khả năng đổi việc.
- Không có nhóm training nào vượt trội => chương trình training không đủ tác động, hoặc quá chung chung.

=> Training Hours **KHÔNG** phải yếu tố rủi ro lớn.

## **8. Nhận xét tổng quan dữ liệu**

- Một số cột **categorical** có tỷ lệ thiếu khá cao như `gender` (~23%), `major_discipline` (~15%), `company_size` (~31%) và `company_type` (~32%), cho thấy thông tin hồ sơ ứng viên không đầy đủ và cần xử lý missing ở bước preprocessing.

- Ở nhóm **numeric**, đa phần dữ liệu sạch, ngoại trừ `training_hours` có khá nhiều outlier (~5.7%) và phân phối bị lệch mạnh về phía các giá trị lớn. `city_development_index` tập trung cao, phản ánh đa số ứng viên ở khu vực phát triển. `experience` và `training_hours` đều mang tính phân tán lớn, cần chuẩn hóa trước modeling.

- Một số **categorical** phân phối không đồng đều, ví dụ `gender` nghiêng mạnh về **Male**; `city`, `education_level`, `enrolled_university` đều bị chi phối bởi một số nhóm giá trị chính, thể hiện sự mất cân bằng tự nhiên trong dữ liệu.

- Ngoại trừ `training_hours`, các biến **numeric** khác hầu như không có outlier đáng kể. Tuy nhiên, phân phối lệch và sự không cân bằng giữa các nhóm giá trị cho thấy dữ liệu cần được chuẩn hoá và xử lý lại trước khi đưa vào mô hình.
