### Dataset Introduction
This dataset (attached) contains information about **baby names** registered in California (USA) from 1910 to 2022.

#### Data Structure:
- **State**: State (CA = California)
- **Sex**: Gender (F = Female, M = Male)
- **Year**: Year of birth
- **Name**: Name
- **Count**: The number of babies with that name in that year

#### Example:
```
CA,F,2020,Olivia,2353  → In 2020, there were 2,353 baby girls named Olivia
CA,M,2020,Noah,2458    → In 2020, there were 2,458 baby boys named Noah
```

### Task 3: Analysis by Decade (1910s, 1920s, ..., 2020s)
**Instruction**
- Calculate the total number of babies born in each decade (create Decade column).
- Which decade had the highest number of babies?
- In the 1990s decade, how many unique names were there?
- "Very Popular" name mean Count >= 500. Compare the number of "Very Popular" names between the 1980s and 2010s decades.
- Calculate the average Count for each decade.

````python
#TODO: your code here

# Calculate the total number of babies born in each decade (create Decade column).
df['Decade'] = (df['Year'] // 10) * 10
babies_of_decade = {}
for index, rows in df.iterrows():
    decade = rows['Decade']
    count = rows['Count']
    if decade in babies_of_decade:
        babies_of_decade[decade] += count
    else:
        babies_of_decade[decade] = count

# Which decade had the highest number of babies?
decade_highest_babies = max(babies_of_decade, key=babies_of_decade.get)

# In the 1990s decade, how many unique names were there?
decade_1990_records = df[df['Decade'] == 1990]
unique_names_1990 = decade_1990_records['Name'].unique()

# "Very Popular" name mean Count >= 500. Compare the number of "Very Popular" names between the 1980s and 2010s decades.
def very_popular_name(decade_info):
    count = 0
    for index, rows in decade_info.iterrows():
        if rows['Count'] >= 500:
            count += 1
        else:
            count += 0
    return count
very_popular_1980 = very_popular_name(df[df['Decade'] == 1980])
very_popular_2010 = very_popular_name(df[df['Decade'] == 2010])

# Calculate the average Count for each decade.
average_count_of_decade = {}
len_of_decade = {}
for index, rows in df.iterrows():
    decade = rows['Decade']
    if decade in len_of_decade:
        len_of_decade[decade] += 1
    else:
        len_of_decade[decade] = 1
for decade in babies_of_decade:
    total_babies = babies_of_decade[decade]
    num_names = len_of_decade[decade]
    average = total_babies / num_names
    average_count_of_decade[decade] = int(average)
````

````python
#TODO: your code here

# Calculate the total number of babies born in each decade (create Decade column).
print(f'The total number of babies born in each decade:\n{babies_of_decade}')

# Which decade had the highest number of babies?
print(f'Decade {decade_highest_babies} had the highest number of babies with {babies_of_decade[decade_highest_babies]} babies.')

# In the 1990s decade, how many unique names were there?
print(f'\nIn the 1990s decade, there are {len(unique_names_1990)} uniques names.')

# "Very Popular" name mean Count >= 500. Compare the number of "Very Popular" names between the 1980s and 2010s decades.
print(f'\nThe number of "Very Popular" names in the 1980s is {very_popular_1980} and 2010s is {very_popular_2010}.')
print('So, in 2010s, there are more \'Very Popular\' names than in 1980s.')

# Calculate the average Count for each decade.
print(f'The average Count for each decade:\n{average_count_of_decade}')
````

### Task 4: Analyze `Timeless` Names (Unchanging Over Time)
**Definition**: A name is called `timeless` if it:
- Appeared in **ALL** decades from the 1910s to the 2020s
- Always had a Count $\ge$ 100 in every decade

**Instruction**
- Find all "timeless" names.
- How many such names are there?
- Display the list of these names.
- Among these names, which one has the highest total Count?

````python
#TODO: your code here

# Find all "timeless" names
all_decades = df['Decade'].unique()
def check_timeless(group):
    is_high_count = (group['Count'] >= 100).all()
    is_in_all_decades = len(group) == len(all_decades)
    return is_high_count and is_in_all_decades
df_timeless = df.groupby('Name').filter(check_timeless)
timeless_name_list = df_timeless['Name'].unique().tolist()

# How many such names are there?
num_timeless_names = len(timeless_name_list)

# Among these names, which one has the highest total Count?
if num_timeless_names == 0:
    highest_total_count_name = 'N/A'
    highest_total_count = 0
else:
    total_count_timeless = df_timeless.groupby('Name')['Count'].sum()
    highest_total_count_name = total_count_timeless.idxmax()
    highest_total_count = total_count_timeless.max()
````

````python
# How many such names are there?
print(f'There are {num_timeless_names} timeless names.')

# Display the list of these names.
print(f'\nThese names are: {timeless_name_list}')

# Among these names, which one has the highest total Count?
print(f'\nAmong these names, {highest_total_count_name} has the highest total Count with {highest_total_count}')
````

### Task 5: `Your Name` Analysis
**Perform an in-depth analysis of any given name**\
Create a function `analyze_name(name)` that performs the following:

1. Check if the name exists in the dataset.
2. If it exists, display:
   - The first year it appeared.
   - The most popular year (and its Count).
   - The total number of people with this name.
   - Which gender uses this name more frequently.
   - Which category this name belongs to (Rare/Uncommon/Common/Very Popular) in the last 5 years.
     
***Test with these name:***
```
['Emma', 'Noah', 'Michael', 'Sophia', 'James']
```

````python
#TODO: your code here

def analyze_name(name, df):
    name_data = df[df['Name'] == name].copy()
    if name_data.empty:
        print(f'{name} is not exist in dataset.')
        return
    # The first year it appeared.
    first_year = name_data['Year'].min()
    print(f'- The first year it appeared: {first_year}')
    
    # The most popular year (and its Count).
    idx_max = name_data['Count'].idxmax()
    most_popular_record = name_data.loc[idx_max]
    most_popular_year = most_popular_record['Year']
    max_count = most_popular_record['Count']
    print(f'- The most popular year (and its Count): {most_popular_year} with {max_count} times')

    # The total number of people with this name.
    total_count = name_data['Count'].sum()
    print(f'- The total number of people with this name: {total_count}')

    # Which gender uses this name more frequently.
    gender_usage = name_data.groupby('Sex')['Count'].sum()
    most_frequent_gender = gender_usage.idxmax()
    print(f'- Which gender uses this name more frequently: {most_frequent_gender}')

    # Which category this name belongs to (Rare/Uncommon/Common/Very Popular) in the last 5 years.
    current_year = df['Year'].max()
    last_5_years = name_data[name_data['Year'] >= current_year - 4]
    last_5_year_total_count = last_5_years['Count'].sum()
    if last_5_year_total_count >= 500:
        category = "Very Popular"
    elif last_5_year_total_count >= 100 and last_5_year_total_count < 500:
        category = "Common"
    elif last_5_year_total_count >= 10 and last_5_year_total_count < 100:
        category = "Uncommon"
    else:
        category = "Rare"
    print(f'- Which category this name belongs to in the last 5 years: {category}')
````

````python
for name in ['Emma', 'Noah', 'Michael', 'Sophia', 'James']:
    print(f'{name}:')
    analyze_name(name, df)
    print('----------\n')
````
